## Topic 2 · Parameters and arguments
*Walkthrough: `02 - Parameters and Arguments`*

### D5. Arguments are matched by position

1. Define `within_limit(value, limit)` that returns `value <= limit`.
2. Print `within_limit(87, 100)`, then `within_limit(100, 87)`.
3. Add a comment explaining why the two results are different.

**Expected output:** `True`, then `False`

````python
# D5
def within_limit(value, limit):
    return value <= limit

print(within_limit(87, 100))
print(within_limit(100, 87))
# The results differ because arguments are matched by position: the first
# argument always goes to value and the second to limit. Swapping them swaps
# the comparison: 87 <= 100 is True, but 100 <= 87 is False.
````

### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:
   - `"OVER LIMIT"` if `percent` is 100 or more
   - `"WARNING"` if `percent` is `warning_at` or more
   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
def status_of(percent, warning_at=90):
    if percent >= 100:
        return "OVER LIMIT"
    elif percent >= warning_at:
        return "WARNING"
    else:
        return "OK"

print(status_of(95))
print(status_of(85))
print(status_of(85, 80))
````

### D8. `*args` and `**kwargs`

a) Define `total(*numbers)` that returns the sum of all the numbers passed in. Print `total(1, 2, 3)`,
   `total(10, 20, 30, 40)` and `total()`.

b) Define `configure(**settings)` that loops over `settings.items()` and prints each one as `key = value`.
   Call it with `configure(host="10.0.0.1", port=8080)`.

**Expected output:**
```
6
100
0
host = 10.0.0.1
port = 8080
```

````python
# D8
def total(*numbers):
    """Return the sum of all the numbers passed in."""
    answer = 0
    for number in numbers:
        answer = answer + number
    return answer

print(total(1, 2, 3))
print(total(10, 20, 30, 40))
print(total())

def configure(**settings):
    """Print each named setting as key = value."""
    for key, value in settings.items():
        print(key, "=", value)

configure(host="10.0.0.1", port=8080)
````

### D9. Local means local

The code is already in the cell below.

1. **Before you run it**, write a comment predicting what will happen.
2. Run it and read the last line of the error.
3. Add a comment explaining why `result` cannot be seen outside the function.

**Expected output:** `inside: OK`, then `NameError: name 'result' is not defined`

````python
# D9
# Prediction: "inside: OK" will print, then the last line will crash with a
# NameError because result is not defined outside the function.
def make_status():
    result = "OK"
    print("inside:", result)

make_status()
print("outside:", result)

# Why result cannot be seen outside: result is created inside make_status(), so
# it is a local variable. It exists only while the function runs and is thrown
# away when the function ends, so the main program has no variable called result.
````

### D10. Parameters are local too

1. **Before you run it**, predict the two numbers it will print.
2. Run it, then add a comment explaining why `number` is still 21.

**Expected output:** `42`, then `21`

````python
# D10
# Prediction: the first print shows 42 (21 * 2) and the second shows 21.
def double(n):
    n = n * 2
    return n

number = 21
print(double(number))
print(number)

# number is still 21 because the parameter n is a local variable inside double().
# n receives a copy of the value 21, and changing n only changes the local copy.
# The outside variable number is never touched.
````

### D12. Returning more than one value

1. Define `check_record(value, limit)` that works out `difference = value - limit` and
   `percent = (value / limit) * 100`, and returns **both**: `return difference, percent`
2. Call it with `check_record(87.0, 100.0)` and store the results in two variables.
3. Print them.

**Expected output:**
```
Difference: -13.0
Percent: 87.0
```

````python
# D12
def check_record(value, limit):
    """Return the difference and the percentage of value against limit."""
    difference = value - limit
    percent = (value / limit) * 100
    return difference, percent

difference, percent = check_record(87.0, 100.0)
print("Difference:", difference)
print("Percent:", percent)
````

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |
| `count_over()` | `0` |

````python
# C2
def count_over(*values, limit=100):
    """Return how many of the values are over the limit."""
    count = 0
    for value in values:
        if value > limit:
            count = count + 1
    return count

print(count_over(87, 120, 45, 101))
print(count_over(87, 120, 45, 101, limit=50))
print(count_over())
````
